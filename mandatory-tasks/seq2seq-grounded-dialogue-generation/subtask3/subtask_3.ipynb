Actually Building Conversational examples

```python
import pandas as pd

def build_examples(split):

    rows = []

    for i in range(len(ds[split])):
        row = ds[split][i]

        rows.append({
            "uid": row["uid"],
            "user2_id": row["user2_id"],
            "wiki_id": row["wikiDocumentIdx"],
            "timestamp": row["utcTimestamp"],
            "hinglish": row["translation"]["hi_en"]
        })

    df = pd.DataFrame(rows)

    # Put turns in conversation order
    df = df.sort_values(
        ["uid", "user2_id", "wiki_id", "timestamp"]
    )

    examples = []

    for _, group in df.groupby(
        ["uid", "user2_id", "wiki_id"]
    ):

        turns = group["hinglish"].tolist()
        wiki_id = group["wiki_id"].iloc[0]

        document = wiki_documents[wiki_id]

        # Keep document manageable
        document = " ".join(
            document.split()[:120]
        )

        for i in range(1, len(turns)):

            history = turns[max(0, i-3):i]
            target = turns[i]

            examples.append({
                "history": " ".join(history),
                "document": document,
                "target": target
            })

    return pd.DataFrame(examples)
```

```python
train_examples = build_examples("train")
val_examples = build_examples("validation")
test_examples = build_examples("test")

print("Train:", len(train_examples))
print("Validation:", len(val_examples))
print("Test:", len(test_examples))
```

TOKENIZER

```python
import re

def tokenize_hinglish(text):
    text = text.lower()
    text = re.sub(r"([?.!,])", r" \1", text)
    return text.split()
```

```python
MAX_HISTORY_LEN = 30
MAX_DOCUMENT_LEN = 80
MAX_TARGET_LEN = 30
MAX_VOCAB_SIZE = 8000

PAD = 0
SOS = 1
EOS = 2
UNK = 3
```

Building voacb from training data

```python
from collections import Counter

counter = Counter()

for _, row in train_examples.iterrows():

    counter.update(
        tokenize_hinglish(row["history"])
    )

    counter.update(
        tokenize_hinglish(row["document"])
    )

    counter.update(
        tokenize_hinglish(row["target"])
    )

most_common = counter.most_common(MAX_VOCAB_SIZE - 4)

vocab = {
    "<PAD>": PAD,
    "<SOS>": SOS,
    "<EOS>": EOS,
    "<UNK>": UNK
}

for i, (word, count) in enumerate(most_common, start=4):
    vocab[word] = i

idx_to_word = {
    index: word
    for word, index in vocab.items()
}

print("Vocabulary size:", len(vocab))
```

Converting Sentences to IDs (same as LSTM/RNN workflow)

```python
def text_to_ids(text, max_len):

    tokens = tokenize_hinglish(text)

    tokens = tokens[:max_len - 2]

    ids = [SOS]

    for token in tokens:
        ids.append(vocab.get(token, UNK))

    ids.append(EOS)

    return ids
```

2 LSTM ENCODERS

```python
import torch

device = torch.device(
    "cuda" if torch.cuda.is_available() else "cpu"
)

print("Device:", device)
```

```python
import torch.nn as nn

EMBEDDING_DIM = 128
HIDDEN_DIM = 128

class Encoder(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim):
        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size,
            embedding_dim,
            padding_idx=PAD
        )

        self.lstm = nn.LSTM(
            embedding_dim,
            hidden_dim,
            batch_first=True
        )

    def forward(self, x):
        embedded = self.embedding(x)

        outputs, (hidden, cell) = self.lstm(embedded)

        return outputs, hidden, cell
```

```python
history_encoder = Encoder(
    len(vocab),
    EMBEDDING_DIM,
    HIDDEN_DIM
).to(device)

document_encoder = Encoder(
    len(vocab),
    EMBEDDING_DIM,
    HIDDEN_DIM
).to(device)
```

Implementing Luong Attention

```python
class LuongAttention(nn.Module):
    def __init__(self):
        super().__init__()

    def forward(self, decoder_hidden, encoder_outputs, mask):

        # decoder_hidden:
        # (1, batch, hidden)

        query = decoder_hidden[-1].unsqueeze(2)

        # encoder_outputs:
        # (batch, sequence_length, hidden)

        scores = torch.bmm(
            encoder_outputs,
            query
        ).squeeze(2)

        # Ignore padding positions
        scores = scores.masked_fill(
            mask == 0,
            -1e9
        )

        weights = torch.softmax(
            scores,
            dim=1
        )

        return weights
```

1 DECODER

```python
class DialogueDecoder(nn.Module):
    def __init__(self, vocab_size, embedding_dim, hidden_dim):
        super().__init__()

        self.embedding = nn.Embedding(
            vocab_size,
            embedding_dim,
            padding_idx=PAD
        )

        self.attention = LuongAttention()

        self.lstm = nn.LSTM(
            embedding_dim + hidden_dim,
            hidden_dim,
            batch_first=True
        )

        self.output_layer = nn.Linear(
            hidden_dim,
            vocab_size
        )

    def forward(
        self,
        x,
        hidden,
        cell,
        encoder_outputs,
        mask
    ):

        embedded = self.embedding(x)

        attention_weights = self.attention(
            hidden,
            encoder_outputs,
            mask
        )

        context = torch.bmm(
            attention_weights.unsqueeze(1),
            encoder_outputs
        )

        lstm_input = torch.cat(
            [embedded, context],
            dim=2
        )

        output, (hidden, cell) = self.lstm(
            lstm_input,
            (hidden, cell)
        )

        prediction = self.output_layer(output)

        return prediction, hidden, cell
```

```python
decoder = DialogueDecoder(
    len(vocab),
    EMBEDDING_DIM,
    HIDDEN_DIM
).to(device)
```

```python
from torch.utils.data import Dataset, DataLoader
from torch.nn.utils.rnn import pad_sequence

class DialogueDataset(Dataset):

    def __init__(self, dataframe):
        self.data = dataframe

    def __len__(self):
        return len(self.data)

    def __getitem__(self, index):

        row = self.data.iloc[index]

        history = torch.tensor(
            text_to_ids(
                row["history"],
                MAX_HISTORY_LEN
            )
        )

        document = torch.tensor(
            text_to_ids(
                row["document"],
                MAX_DOCUMENT_LEN
            )
        )

        target = torch.tensor(
            text_to_ids(
                row["target"],
                MAX_TARGET_LEN
            )
        )

        return history, document, target
```

```python
def collate_fn(batch):

    histories = [item[0] for item in batch]
    documents = [item[1] for item in batch]
    targets = [item[2] for item in batch]

    histories = pad_sequence(
        histories,
        batch_first=True,
        padding_value=PAD
    )

    documents = pad_sequence(
        documents,
        batch_first=True,
        padding_value=PAD
    )

    targets = pad_sequence(
        targets,
        batch_first=True,
        padding_value=PAD
    )

    return histories, documents, targets
```

```python
from torch.utils.data import DataLoader

train_small = train_examples.iloc[:10000].reset_index(drop=True)

train_dataset = DialogueDataset(train_small)
val_dataset = DialogueDataset(val_examples)
test_dataset = DialogueDataset(test_examples)

train_loader = DataLoader(
    train_dataset,
    batch_size=64,
    shuffle=True,
    collate_fn=collate_fn
)

val_loader = DataLoader(
    val_dataset,
    batch_size=64,
    shuffle=False,
    collate_fn=collate_fn
)

test_loader = DataLoader(
    test_dataset,
    batch_size=64,
    shuffle=False,
    collate_fn=collate_fn
)

print("Train batches:", len(train_loader))
print("Validation batches:", len(val_loader))
print("Test batches:", len(test_loader))
```

Training Loop

```python
import time

def train_epoch():

    history_encoder.train()
    document_encoder.train()
    decoder.train()

    total_loss = 0

    for history, document, target in train_loader:

        history = history.to(device)
        document = document.to(device)
        target = target.to(device)

        history_optimizer.zero_grad()
        document_optimizer.zero_grad()
        decoder_optimizer.zero_grad()

        # Encode history
        history_outputs, history_hidden, history_cell = history_encoder(history)

        # Encode document
        document_outputs, document_hidden, document_cell = document_encoder(document)

        # Combine both encoder outputs
        encoder_outputs = torch.cat(
            [history_outputs, document_outputs],
            dim=1
        )

        # Padding mask
        history_mask = history != PAD
        document_mask = document != PAD

        mask = torch.cat(
            [history_mask, document_mask],
            dim=1
        )

        # Combine encoder hidden states
        hidden = (history_hidden + document_hidden) / 2
        cell = (history_cell + document_cell) / 2

        decoder_input = target[:, :1]

        loss = 0

        for t in range(1, target.size(1)):

            output, hidden, cell = decoder(
                decoder_input,
                hidden,
                cell,
                encoder_outputs,
                mask
            )

            loss += criterion(
                output.squeeze(1),
                target[:, t]
            )

            # Teacher forcing
            decoder_input = target[:, t:t+1]

        loss = loss / (target.size(1) - 1)

        loss.backward()

        history_optimizer.step()
        document_optimizer.step()
        decoder_optimizer.step()

        total_loss += loss.item()

    return total_loss / len(train_loader)
```

Generating Hinglish response

```python
def generate_response(history_text, document_text):

    history_encoder.eval()
    document_encoder.eval()
    decoder.eval()

    history = torch.tensor(
        text_to_ids(history_text, MAX_HISTORY_LEN)
    ).unsqueeze(0).to(device)

    document = torch.tensor(
        text_to_ids(document_text, MAX_DOCUMENT_LEN)
    ).unsqueeze(0).to(device)

    with torch.no_grad():

        history_outputs, history_hidden, history_cell = history_encoder(history)

        document_outputs, document_hidden, document_cell = document_encoder(document)

        encoder_outputs = torch.cat(
            [history_outputs, document_outputs],
            dim=1
        )

        history_mask = history != PAD
        document_mask = document != PAD

        mask = torch.cat(
            [history_mask, document_mask],
            dim=1
        )

        hidden = (history_hidden + document_hidden) / 2
        cell = (history_cell + document_cell) / 2

        decoder_input = torch.tensor(
            [[SOS]],
            device=device
        )

        words = []

        for _ in range(MAX_TARGET_LEN):

            output, hidden, cell = decoder(
                decoder_input,
                hidden,
                cell,
                encoder_outputs,
                mask
            )

            next_token = output[:, -1, :].argmax(1).item()

            if next_token == EOS:
                break

            if next_token != PAD:
                words.append(
                    idx_to_word.get(next_token, "<UNK>")
                )

            decoder_input = torch.tensor(
                [[next_token]],
                device=device
            )

    return " ".join(words)
```

```python
def build_examples(split):

    rows = []

    for i in range(len(ds[split])):

        row = ds[split][i]

        rows.append({
            "uid": row["uid"],
            "user2_id": row["user2_id"],
            "wiki_id": row["wikiDocumentIdx"],
            "doc_idx": row["docIdx"],
            "timestamp": row["utcTimestamp"],
            "hinglish": row["translation"]["hi_en"]
        })

    df = pd.DataFrame(rows)

    df = df.sort_values(
        ["uid", "user2_id", "wiki_id", "timestamp"]
    )

    examples = []

    for _, group in df.groupby(
        ["uid", "user2_id", "wiki_id"]
    ):

        turns = group["hinglish"].tolist()
        wiki_id = group["wiki_id"].iloc[0]
        doc_idx = group["doc_idx"].iloc[0]

        # Get only the relevant Wikipedia section
        filename = wiki_mapping[wiki_id]

        with open(
            "datasets-CMU_DoG/WikiData/" + filename,
            "r"
        ) as f:
            wiki_data = json.load(f)

        section = wiki_data[str(doc_idx)]

        # Convert section into text
        if isinstance(section, dict):

            section_parts = []

            for key, value in section.items():

                if isinstance(value, list):
                    value = " ".join(value)

                section_parts.append(
                    key + ": " + str(value)
                )

            document = " ".join(section_parts)

        else:
            document = str(section)

        # Keep document manageable
        document = " ".join(
            document.split()[:120]
        )

        # Create history -> target examples
        for i in range(1, len(turns)):

            history = turns[max(0, i-3):i]
            target = turns[i]

            examples.append({
                "history": " ".join(history),
                "document": document,
                "target": target
            })

    return pd.DataFrame(examples)
```

```python
train_examples = build_examples("train")
val_examples = build_examples("validation")
test_examples = build_examples("test")

print("Train:", train_examples.shape)
print("Validation:", val_examples.shape)
print("Test:", test_examples.shape)
```
